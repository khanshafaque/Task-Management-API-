```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import plotly.express as px
import matplotlib.pyplot as plt

np.random.seed(0)

data = {
    'Product': ['Product A', 'Product B', 'Product C', 'Product D', 'Product E'],
    'Sales': np.random.randint(50, 500, size=5),
    'Profit': np.random.randint(20, 200, size=5),
    'Year': [2020, 2021, 2020, 2021, 2020],
    'QuantitySold': np.random.randint(1, 100, size=5)
}

df = pd.DataFrame(data)
```

```python
 plt.figure(figsize=(8,6))
 plt.bar(df['Product'],df['Sales'],color='skyblue')
 plt.title('Sales by Product')
 plt.xlabel('Product')
 plt.ylabel('Sales')
 plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.plot(df['Product'], df['Sales'], marker='o', linestyle='-', color='green', label='Sales')
plt.plot(df['Product'], df['Profit'], marker='s', linestyle='-', color='orange', label='Profit')
plt.title('Sales vs Profit by Product')
plt.xlabel('Product')
plt.ylabel('Amount')
plt.legend()
plt.show()
```

```python
 plt.figure(figsize=(8,6))
 sns.scatterplot(x='Sales',y='Profit',data=df,hue='Product',palette='deep',s=100)
 plt.title('Sales vs Profit (scatter plot)')
 plt.xlabel('sales')
 plt.ylabel('Profit')
 plt.show()
```

```python
 plt.figure(figsize=(8,6))
 sns.boxplot(x='Year',y='Sales',data=df,palette='Blues')
 plt.title('Sales Distribution by Year')
 plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.hist(df['Sales'], bins=10, color='lightcoral', edgecolor='black')
plt.title('Distribution of Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
 df['ProfitMargin']=df['Profit']/df['Sales']*100 #creating new column
 correlation_matrix=df[['Sales','Profit','QuantitySold','ProfitMargin']].corr()
 plt.figure(figsize=(8,6))
 sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm',fmt='.2f',linewidths=1)
 plt.title('Correlation Heatmap')
 plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.pie(
    df['Sales'],
    labels=df['Product'],
    autopct='%1.1f%%',
    startangle=90,
    colors=['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0']
)
plt.title('Sales Distribution by Product')
plt.show()
```

```python
 fig=px.scatter(df, x='Sales', y='Profit', color='Product', size='QuantitySold',title='Sales vs Profit (Interactive Plot)', labels={'Sales': 'Sales Amount', 'Profit': 'Profit Amount'})
 fig.show()
```

```python
import plotly.express as px

fig = px.bar(
    df,
    x='Product',
    y='Sales',
    color='Product',
    title='Sales by Product (Interactive Bar Chart)'
)
fig.show()
```
